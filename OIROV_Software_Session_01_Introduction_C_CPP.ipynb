# OIROV Software Track — Session 01
## Introduction to Programming (C/C++ foundations)

**Goal:** understand programming foundations used later with Arduino. Arduino sketches use C++ with Arduino functions. Standard C examples are explanatory; Arduino sketches should be copied into Arduino IDE, not run as Python cells.

Topics: algorithms, variables, operators, conditions, loops, functions, setup()/loop(), and Blink.

## 1. Program and algorithm
A **program** is a set of instructions. An **algorithm** is an ordered set of steps used to solve a problem.

Example: start → turn LED on → wait → turn LED off → wait → repeat.

## 2. First C program

```c
#include <stdio.h>

int main(void) {
    printf("Hello, OIROV!\n");
    return 0;
}
```

- `stdio.h` provides `printf()`.
- `main()` is the entry point of a standard C program.
- `return 0;` indicates successful completion.

Expected output: `Hello, OIROV!`

## 3. Variables and data types

```c
int studentCount = 12;
float batteryVoltage = 7.4;
char teamInitial = 'O';

printf("Students: %d\n", studentCount);
printf("Battery voltage: %.1f V\n", batteryVoltage);
printf("Team initial: %c\n", teamInitial);
```

- `int`: whole numbers
- `float`: decimal numbers
- `char`: one character

## 4. Operators

```c
int left = 10, right = 3;
printf("Addition: %d\n", left + right);
printf("Subtraction: %d\n", left - right);
printf("Multiplication: %d\n", left * right);
printf("Integer division: %d\n", left / right);
printf("Remainder: %d\n", left % right);
printf("left > right: %d\n", left > right);
```

In C, integer division `10 / 3` gives `3`. A true comparison prints `1`; false prints `0`.

## 5. Conditions: if / else

```c
int flameDetected = 1;
if (flameDetected == 1) {
    printf("Flame detected!\n");
} else {
    printf("No flame detected.\n");
}
```

This is a simplified logic example. Test the actual sensor and threshold before using them in a robot.

## 6. Loops

### for loop
```c
for (int count = 1; count <= 5; count++) {
    printf("Blink number: %d\n", count);
}
```

### while loop
```c
int count = 1;
while (count <= 3) {
    printf("Step: %d\n", count);
    count++;
}
```
Loops repeat instructions, useful when a robot repeats a task or continuously checks a condition.

## 7. Functions

```c
void showReadyMessage(void) {
    printf("OIROV robot is ready!\n");
}

int main(void) {
    showReadyMessage();
    return 0;
}
```

A function groups instructions under a name so they can be reused.

## 8. Arduino program structure

```cpp
void setup() {
    // Runs once when the board starts or resets
}

void loop() {
    // Repeats while the board is powered
}
```

`setup()` initializes the board; `loop()` repeats continuously. Comments after `//` are ignored by the compiler.

## 9. First Arduino program: Blink
Copy this sketch into **Arduino IDE** and select the correct board and port before uploading.

```cpp
void setup() {
    pinMode(LED_BUILTIN, OUTPUT);
}

void loop() {
    digitalWrite(LED_BUILTIN, HIGH); // LED on
    delay(1000);                     // wait about 1 second
    digitalWrite(LED_BUILTIN, LOW);  // LED off
    delay(1000);                     // wait about 1 second
}
```

Change `delay(1000)` to `delay(500)` for faster blinking or `delay(2000)` for slower blinking. `LED_BUILTIN` is preferred over hard-coding a pin number. This is an Arduino sketch, not a standard C program for direct execution in Jupyter.
