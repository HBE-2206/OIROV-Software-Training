## 1. What makes embedded software different?
Embedded software runs on a device that interacts with physical hardware and often has limited memory, timing, and power. A bug can affect motors, pumps, or other hardware, so predictable behavior and safe defaults matter.

## 2. Arduino sketch structure
```cpp
const int ledPin = LED_BUILTIN;

void setup() {
  pinMode(ledPin, OUTPUT);
  Serial.begin(9600);
  Serial.println("System started");
}

void loop() {
  // Repeated application logic goes here.
}
```
- Constants document fixed configuration.
- `setup()` runs once after reset/start.
- `loop()` runs repeatedly.
- `pinMode()` configures a digital pin; `Serial` helps inspect behavior.

## 3. Blocking delay vs elapsed-time timing
`delay()` is useful for the first Blink exercise, but it pauses the loop while waiting. During that pause, other tasks may not be checked promptly.

```cpp
const unsigned long interval = 1000;
unsigned long previousMillis = 0;

void setup() {
  pinMode(LED_BUILTIN, OUTPUT);
}

void loop() {
  unsigned long now = millis();
  if (now - previousMillis >= interval) {
    previousMillis = now;
    digitalWrite(LED_BUILTIN, !digitalRead(LED_BUILTIN));
  }
}
```
This example toggles an LED using elapsed time. Use unsigned time variables and subtraction-based comparisons for rollover-safe interval checks.

## 4. State-based program design
Represent the robot's behavior as explicit states, for example:
- `IDLE` — outputs are in safe state.
- `SEARCHING` — robot follows the approved search behavior.
- `TARGET_CHECK` — validate sensor evidence.
- `EXTINGUISHING` — activate the approved mechanism under controlled conditions.
- `FAULT` — stop motion and disable outputs where appropriate.

Transitions should have clear conditions. This is a design outline, not a complete safety controller.

## 5. Debugging checklist
1. Confirm the selected board and port.
2. Compile before uploading and read the first meaningful error.
3. Confirm pin numbers and board-specific definitions.
4. Add Serial messages around important decisions.
5. Test one subsystem at a time.
6. Record the expected and actual behavior.
7. Remove or reduce noisy debug output once timing matters.

## 7. Arduino Uno vs ESP32 notes
- Select the exact board in Arduino IDE before compiling.
- Pin labels, available peripherals, ADC characteristics, and voltage limits differ.
- Do not treat 5 V Uno logic and 3.3 V ESP32 GPIO as interchangeable.
- Check board documentation and sensor interface requirements before wiring.
- Keep code portable where practical, but isolate board-specific pins and features.
