## 1. A repeatable problem-solving method
1. State the task in one sentence.
2. Identify inputs and expected outputs.
3. Write the steps in plain language or pseudocode.
4. Split the solution into small functions.
5. Test normal, boundary, and failure cases.
6. Record the result and what changed.

## 2. Function anatomy
A function gives a name to a reusable operation. Parameters carry input into it; a return value sends a result back.

```cpp
int addValues(int a, int b) {
  return a + b;
}

void printReadyMessage() {
  Serial.println("OIROV system ready");
}
```

- `int` means the first function returns an integer.
- `void` means the second function does not return a value.
- `a` and `b` are parameters; `addValues(2, 3)` is a call.

## 3. Decompose a robot task
For a future firefighting robot, separate the behavior into responsibilities:
- `readFlameSensor()` — read a sensor and return a measurement/state.
- `isFlameDetected(value)` — compare a reading against a calibrated threshold.
- `stopMotors()` — request a safe stop through the motor driver.
- `activatePump()` — command the pump driver only when conditions permit.

These are design examples, not complete production functions. Pin mappings, driver circuitry, thresholds, and emergency-stop behavior must be defined and tested for the actual hardware.

## 4. Worked example: classify a sensor reading
```cpp
bool isFlameDetected(int reading, int threshold) {
  return reading >= threshold; // Example only: sensor polarity varies
}

void setup() {
  Serial.begin(9600);
  int sample = 720;
  Serial.println(isFlameDetected(sample, 600) ? "DETECTED" : "NOT DETECTED");
}

void loop() {
  // No repeated action needed in this small demonstration.
}
```
**Important:** the comparison direction is only illustrative. Check the sensor datasheet and collect real readings before choosing a threshold.

## 5. Scope and naming
- Declare variables close to where they are needed.
- Use descriptive names such as `sensorReading`, not `x1`.
- Avoid changing global variables from many functions without a clear reason.
- Prefer one clear responsibility per function.
- Keep hardware actions separate from decision logic where practical.

## 6. Practical lab (45 minutes)
**Task:** write a small decision program for a robot status.
1. Create `bool isLowBattery(float voltage, float limit)`.
2. Create `void printStatus(float voltage)` that prints `LOW BATTERY` or `BATTERY OK`.
3. Test values below, equal to, and above the chosen limit.
4. Add comments explaining the decision.
5. Write down at least three test cases and their expected outputs.

**Acceptance criteria:** functions have clear names, comparison is correct for the chosen rule, boundary cases are tested, and output is documented.

## 7. Review questions

2. What is the difference between a parameter and an argument?
3. Why test a value exactly equal to the threshold?
4. Name two functions that could separate sensing from motor control.
