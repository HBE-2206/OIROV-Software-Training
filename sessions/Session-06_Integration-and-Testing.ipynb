**Trainer:** Hazem Elshafey — CTO, OIROV Robotics
**Track:** Software | **Duration:** 2 hours | **Program:** 3 weeks, 2 sessions per week

**Session goal:** integrate sensor, decision, movement, and extinguishing modules using a staged test plan, clear documentation, and safety checks.

## Learning outcomes
By the end of this session, learners should be able to:
- Explain the core concepts introduced in the lesson.
- Read and modify the provided C/C++ examples.
- Apply the concept to the software of a small robot.
- Document and test their work rather than relying on an untested upload.

## 1. Proposed software architecture
Keep the application divided into understandable modules:
1. **Sensor layer:** reads raw sensor values and reports validity.
2. **Decision layer:** classifies target evidence and selects a state.
3. **Motion layer:** maps commands to motor-driver outputs.
4. **Extinguishing layer:** controls the approved pump/actuator interface.
5. **Safety/state layer:** startup, timeout, fault, and stop behavior.
6. **Diagnostics:** serial logs and test results.

Avoid allowing every module to change every output directly.

## 2. Example high-level behavior
1. Start with outputs in a safe state.
2. Verify required inputs and initialization.
3. Search only while the system is in the correct state.
4. Validate target evidence rather than reacting to a single noisy sample.
5. Stop movement before performing the approved extinguishing action if required by the design.
6. Confirm the target condition and time limits.
7. Return to a safe state or enter `FAULT` when conditions are not met.

This is a planning flow, not a ready-to-deploy autonomous control algorithm. Final behavior depends on the team's hardware and competition rules.

## 3. Test from small to large
- **Unit test:** test a function with known inputs and expected outputs.
- **Module test:** test sensor reading, movement mapping, or pump-driver command separately.
- **Integration test:** verify that module interfaces and state transitions agree.
- **System test:** test the complete robot in a controlled environment.
- **Regression test:** repeat earlier tests after changing code or wiring.

## 4. Test case template
For every test, record:
- Test ID and date
- Requirement or behavior being tested
- Hardware/software version
- Setup and safety precautions
- Input or action
- Expected result
- Actual result
- Pass/Fail
- Evidence (serial log, photo, or observation)
- Issue and next action

A test is not complete just because the code compiles.

## 5. Integration rules
- Agree on pin names, function names, and module interfaces before combining code.
- Keep one known-good version before each major change.
- Integrate one module at a time.
- Compile after every integration step.
- Test with actuators disconnected or safely isolated first when possible.
- Review shared code changes and document assumptions.

## 6. Practical lab (45 minutes)
**Task:** integrate a simulated firefighting sequence before activating hardware.
1. Implement or describe sensor reading, detection decision, motion command, and pump request as separate functions/modules.
2. Use mock values to simulate no target, target detected, invalid reading, and timeout.
3. Verify state transitions for each case.
4. Ensure invalid input and timeout lead to a safe state.
5. Write test cases and record expected vs actual output.
6. Review with a teammate and note unresolved hardware dependencies.

**Acceptance criteria:** modular design, at least four test cases, explicit fault handling, and readable documentation.

## 7. Final Firefighting Robot evaluation
The training evaluation weights are:
- **20%** technical knowledge
- **30%** practical implementation
- **35%** final project
- **15%** teamwork and documentation

Final project capabilities to demonstrate safely:
- Robot movement under control.
- Detection of an approved, safe flame/thermal target or simulation.
- A decision based on sensor input.
- Activation of the pump/extinguishing mechanism through appropriate driver hardware.
- Safe stop/fault behavior.

Required deliverables: README, source code, wiring/pinout, test records, and final-project documentation.

## 8. Safety checklist
- Disconnect power before rewiring.
- Do not drive motors or pumps directly from GPIO pins.
- Use correctly rated motor drivers/ESCs and power distribution.
- Protect electronics and connections from water; test away from live electronics where appropriate.
- Use a safe simulated target or approved controlled setup.
- Keep a physical power disconnect accessible.
- Test motion and pump functions independently before full operation.

## 9. Final review
1. What is the difference between module testing and integration testing?
2. Why should a fault path be tested explicitly?
3. Which documents should accompany the final project?
4. Why is a successful compile not proof of correct robot behavior?
