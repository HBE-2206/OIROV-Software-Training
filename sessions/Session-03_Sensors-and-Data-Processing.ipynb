## 1. Sensor-to-decision pipeline
**Physical world → sensor → electrical signal → microcontroller reading → filtering/calibration → decision → action.**

A reading is not automatically a correct measurement. Wiring, power, sensor range, noise, lighting, distance, and calibration can change the value.

## 2. Digital vs analog inputs
- **Digital input:** interpreted as a HIGH/LOW state, such as a digital flame module output.
- **Analog input:** converted to a numeric reading by the board's ADC; resolution and voltage range depend on the board.
- Arduino Uno and ESP32 do not necessarily use identical ADC ranges or pin behavior. Never assume a value from one board maps directly to the other.

## 3. Arduino input examples
### Digital input
```cpp
const int sensorPin = 2;

void setup() {
  Serial.begin(9600);
  pinMode(sensorPin, INPUT); // Confirm module output and wiring
}

void loop() {
  int state = digitalRead(sensorPin);
  Serial.println(state);
  delay(200);
}
```
### Analog input
```cpp
const int sensorPin = A0;

void setup() {
  Serial.begin(9600);
}

void loop() {
  int reading = analogRead(sensorPin);
  Serial.println(reading);
  delay(200);
}
```
Use the correct pin and input mode for your exact board and sensor. ESP32 ADC-capable pins and allowable voltages differ from Arduino Uno.

## 6. Practical lab (45 minutes)
**Task:** build a sensor observation table before writing a decision rule.
1. Read the supplied digital or analog sensor.
2. Log at least 10 readings with the target absent and 10 with it present.
3. Calculate the approximate range for each condition.
4. Identify overlap or unstable values.
5. Propose a threshold and state its limitations.
6. Test the threshold using several new readings.

**Deliverable:** table of readings, chosen rule, and a short note on false positives/false negatives.

## 7. Common issues
- Constant zero or maximum: check power, ground, pin, and sensor output voltage.
- Random values: check loose wiring, floating inputs, electrical noise, and grounding.
- Logic seems reversed: some modules are active-low.
- Uno code fails on ESP32: verify pin names, ADC range, and board-specific behavior.
- Never expose a microcontroller input to a voltage outside its permitted range.
