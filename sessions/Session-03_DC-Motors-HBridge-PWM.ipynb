## 2. Direction and speed signals
A typical H-Bridge channel uses two direction inputs and an enable/PWM input. The exact truth table depends on the module and wiring. PWM rapidly switches a signal to control average drive; it does not reduce the GPIO voltage to a new fixed voltage.

| Software intent | Typical signal idea | Hardware check |
|---|---|---|
| Forward | IN1/IN2 set to one direction | Confirm motor polarity |
| Reverse | IN1/IN2 reversed | Confirm direction table |
| Stop | Driver-specific stop/coast/brake state | Verify module truth table |
| Speed | PWM on enable input | Confirm enable jumper and PWM-capable pin |

## 3. Illustrative Arduino Uno/L298N sketch
**Do not wire from this example alone.** The pin numbers are placeholders and must match the approved team's wiring plan. Keep motor power disconnected while checking signal wiring.

```cpp
const int enA = 5;   // Example PWM pin; confirm board and wiring
const int in1 = 7;   // Example only
const int in2 = 8;   // Example only

void stopMotor() {
  analogWrite(enA, 0);
  digitalWrite(in1, LOW);
  digitalWrite(in2, LOW);
}

void setup() {
  pinMode(enA, OUTPUT);
  pinMode(in1, OUTPUT);
  pinMode(in2, OUTPUT);
  stopMotor(); // Safe initial software state
}

void loop() {
  // Forward at a conservative test command; test with wheels lifted.
  digitalWrite(in1, HIGH);
  digitalWrite(in2, LOW);
  analogWrite(enA, 100); // Example PWM command, not a calibrated speed
  delay(1000);
  stopMotor();
  delay(1000);
}
```
On ESP32, PWM setup differs by Arduino core/version; check the installed core's supported API rather than assuming Uno `analogWrite()` behavior.

## 4. PWM and command limits
- PWM duty-cycle values and resolution depend on the board/API.
- A command value is not a guaranteed physical speed; load, supply, motor, and driver matter.
- Start with a low command and increase gradually under controlled conditions.
- Avoid sudden direction reversal under load unless the driver and mechanical system are designed for it.
- Check the motor driver's current, voltage, and thermal limits.

## 5. Motor test safety
1. Disconnect power while wiring.
2. Check polarity, common ground requirements, and driver ratings.
3. Secure the chassis and lift wheels clear of the surface.
4. Keep hands and loose wires away from moving parts.
5. Test one motor channel at a time.
6. Keep a physical power disconnect available.
7. Never power a DC motor directly from a microcontroller GPIO pin.

## 6. Practical lab — One motor channel (45 min)
1. With motor supply disconnected, check the control-pin mapping against the team's diagram.
2. Compile the sketch and verify pin configuration.
3. Trainer checks the wiring and power ratings.
4. Apply power only after the safety check; keep wheels lifted.
5. Test stop, one direction, then the other direction at low PWM.
6. Record command, observed direction, and any heating/noise/instability.
7. Stop immediately if the motor or driver behaves unexpectedly.

**Deliverable:** completed motor test table and a confirmed IN1/IN2/EN mapping.
