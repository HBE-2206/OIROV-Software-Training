## 1. Link to Hardware Session 06 — Assembly & Troubleshooting

## 2. Suggested software modules
- `readSensors()` — gather raw sensor inputs.
- `classifyTarget()` — interpret sensor readings using tested polarity/threshold.
- `applyMotionCommand()` — translate intent to verified motor-driver signals.
- `setPump()` — command the approved relay/driver interface.
- `setServoPosition()` — move the servo within its verified mechanical range.
- `enterFaultState()` — report issue and place outputs in defined safe state.

Keep hardware pin mappings in one clearly labeled configuration area. Avoid letting every module write directly to every output.

## 3. Integration order
1. Compile and test the basic Blink sketch.
2. Verify digital input and Serial Monitor output.
3. Test one DC motor channel using the approved L298N pin map with wheels lifted.
4. Test servo positioning without mechanical obstruction.
5. Simulate the pump request with an LED/Serial message; only test the real pump under trainer supervision after power and relay checks.
6. Combine sensor decision and actuator request in a state machine.
7. Run the complete sequence with a safe target or simulation.

## 4. Troubleshooting decision tree
**Sketch won't compile:** confirm board selection, library, API version, and first meaningful compiler error.  
**Upload fails:** confirm cable, port, board, and boot/upload procedure.  
**Sensor stays constant:** check supply, ground, output pin, polarity, and sensor adjustment.  
**Motor does not move:** check motor supply, driver enable, input mapping, common reference, and current rating with power disconnected for wiring changes.  
**Board resets when motor starts:** stop testing and inspect power separation, supply capacity, wiring, grounding, and noise suppression with the hardware lead.  
**Relay/pump behaves reversed:** stop and verify relay active level and contact wiring; do not guess.

## 6. Practical lab — Integrate safely (45 min)
1. Inspect the pinout and power diagram with the hardware lead.
2. Confirm continuity and polarity before applying power.
3. Run each subsystem test separately.
4. Use a simulated target and simulated pump output first.
5. Test at least one invalid-input or timeout case.
6. Record expected vs actual behavior and identify the fault layer: power, wiring, driver, sensor, or software.
7. Combine subsystems only after individual tests pass.

## 7. Final project and assessment
**Final project:** Firefighting Robot. Required demonstration: controlled movement, sensor-based target detection (or approved simulation), a controlled pump/actuator request through suitable driver hardware, and a safe stop/fault path.

**Assessment weights:**

**Deliverables:** source code, README, pinout/wiring diagram, test cases and results, troubleshooting log, and final project notes.

## 8. Final safety checklist
- Disconnect power before changing wiring.
- Confirm measured supply rails and board voltage compatibility.
- Do not power motors or pumps from GPIO pins.
- Keep wheels lifted and secure the robot during motor tests.
- Keep a physical power disconnect accessible.
- Test the pump with a safe controlled setup and protect electronics from water.
- Ask the hardware lead to review any unexplained reset, heating, smoke, or abnormal sound; stop testing immediately.

## Final practical project — Firefighting Robot integration

**Code skeleton:** [Session06_FirefightingRobot_Demo.ino](https://github.com/HBE-2206/OIROV-Software-Training/blob/main/projects/session-06-firefighting-integration/Session06_FirefightingRobot_Demo.ino)

The code is a staged integration demonstration. It starts in a safe idle state, supports a search command from Serial Monitor, checks sensor state, stops motion on target detection, and bounds the pump request. The real pump output is disabled by default.

**Suggested test cases and expected behavior:**
1. Startup → outputs initialized to the configured safe state.
2. Sensor inactive in search → search motor command continues at conservative PWM.
3. Sensor active → motion stops and target confirmation begins.
4. Target confirmed → bounded pump request is issued; by default it is logged as simulation.
5. Operator sends `X` while searching → return to idle.
6. Sensor fails or wiring is uncertain → stop testing; verify hardware and add explicit fault detection before real use.

**Answer key:**
- Unit testing checks a small function; integration testing checks interfaces and modules working together.
- A successful compile does not validate wiring, sensor polarity, mechanical safety, or system behavior.
- Required handover: source code, README, approved pinout/wiring, test cases/results, troubleshooting notes, and demonstration evidence.

**Assessment weights (agreed):** 20% technical knowledge, 30% practical implementation, 35% final project, 15% teamwork/documentation.
