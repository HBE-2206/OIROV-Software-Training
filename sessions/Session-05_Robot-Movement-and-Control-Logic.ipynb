## 1. Safety before motion
- Lift wheels/propellers clear of the ground or water during first tests.
- Secure the chassis and keep hands, hair, and loose wires away from moving parts.
- Test one motor at a time at low power.
- Provide a physical way to disconnect motor power.
- Never power a motor directly from a microcontroller GPIO pin; use a correctly rated driver/ESC and a suitable power source.

## 2. Controller vs motor driver
The microcontroller sends low-power logic signals. A motor driver or ESC switches the motor's higher current from an appropriate supply. Check voltage/current ratings, grounding, protection, cooling, and signal requirements for the exact device before connection.

## 3. Start with movement commands
Design a small set of commands before writing hardware code:
- `STOP`
- `FORWARD`
- `REVERSE`
- `TURN_LEFT`
- `TURN_RIGHT`
- `ASCEND` / `DESCEND` if the vehicle has vertical thrusters

For OIROV's ROV-style layout, horizontal and vertical motors may need separate mappings. Confirm the actual motor count, orientation, polarity, and driver channels against the current wiring plan.

## 4. Conceptual movement mixing
For a simple differential-drive robot:
- Forward: left and right sides move forward.
- Turn left: left side slows/stops while right side moves forward.
- Turn right: right side slows/stops while left side moves forward.
- Stop: all motion commands go to the defined neutral/stop state.

For an ROV with multiple thrusters, movement is a mapping from desired axes to individual motor commands. The mapping depends on the physical layout and should be documented in a table before implementation.

## 5. Separate intent from hardware commands
```cpp
enum MotionCommand {
  STOP,
  FORWARD,
  REVERSE,
  TURN_LEFT,
  TURN_RIGHT
};

void applyMotion(MotionCommand command) {
  // Convert the selected command to driver outputs.
  // Implement only after pinout and driver behavior are verified.
}
```
The enum names intent; `applyMotion()` should own the mapping to the actual hardware. Add input validation and a safe default for unknown commands.

## 6. Stop and fault handling
A robust control loop should consider:
- What happens if controller input disconnects?
- What happens if a command is stale or invalid?
- What is the startup output state?
- How quickly should the system stop after a fault?
- Is the stop action independent of normal movement logic?

Do not rely on software alone when a physical disconnect or emergency stop is required by the test setup.

## 7. Practical lab (45 minutes)
**Task:** build and test a command-to-motor mapping table before connecting real motors.
1. List every motor channel and physical direction.
2. Fill a table for `STOP`, `FORWARD`, `REVERSE`, and turns.
3. Write pseudocode or a simulator that prints each intended output.
4. Add a timeout/failsafe rule for missing commands.
5. Review the mapping with another team member.
6. Only then test one motor channel at low power with the chassis secured.

**Deliverable:** motor map, command table, test observations, and a list of unresolved wiring assumptions.

## 8. Example test matrix
| Test | Input | Expected behavior |
|---|---|---|
| T1 | STOP | All channels reach the approved stop state |
| T2 | FORWARD | Mapped motors move in the intended direction |
| T3 | Invalid command | Safe stop; no undefined output |
| T4 | Command timeout | Stop/fault behavior occurs |
| T5 | Startup/reset | Outputs remain safe until initialized |

Use the actual driver's stop/neutral requirements; zero PWM is not universally equivalent to a safe stop.
