## 1. Link to Hardware Session 02 — Electrical Fundamentals
Hardware topics include voltage, current, polarity, VCC/GND, 3.3 V vs 5 V logic, and using a multimeter. The software-side rule: **a GPIO pin is a signal pin, not a power supply for motors or pumps**. Confirm the permitted input voltage for the exact board and module before connecting them.

## 2. Digital output: LED example
```cpp
const int ledPin = LED_BUILTIN;

void setup() {
  pinMode(ledPin, OUTPUT);
}

void loop() {
  digitalWrite(ledPin, HIGH);
  delay(500);
  digitalWrite(ledPin, LOW);
  delay(500);
}
```
`HIGH` and `LOW` represent logic levels; their electrical voltages depend on the board. `LED_BUILTIN` avoids assuming a particular LED pin.

## 3. Digital input: reading a module output
```cpp
const int sensorOutPin = 2; // Example only: confirm pinout first

void setup() {
  Serial.begin(9600);
  pinMode(sensorOutPin, INPUT);
}

void loop() {
  int state = digitalRead(sensorOutPin);
  Serial.println(state);
  delay(200);
}
```
Some sensor modules are active-low, so a reading of `LOW` may mean detection. Check the module's documentation and test its output; do not assume the polarity.

## 4. Serial Monitor as a debugging tool
- Set `Serial.begin()` to the baud rate selected in Serial Monitor.
- Print the input value before adding complex decisions.
- Print a short label when useful, e.g. `Serial.print("sensor=");`.
- If output is unreadable, check baud rate and USB/serial connection.
- Remove excessive printing when timing becomes important.

## 5. Pin and board differences
- Arduino Uno commonly uses 5 V logic; ESP32 GPIO is generally 3.3 V logic and is not 5 V tolerant unless a specific interface says otherwise.
- ESP32 and Uno pin names, ADC behavior, and PWM APIs can differ.
- Do not copy pin numbers from a diagram without checking the exact board revision and wiring plan.
- Never connect a motor, pump, or relay coil directly to a GPIO pin.
