## 2. What software can and cannot protect
Software can choose safe initial output states, reject invalid commands, time out stale commands, report faults, and avoid starting an actuator before initialization. Software cannot replace a fuse, correctly rated wire, suitable regulator, physical disconnect, flyback protection where required, or safe power design.

## 3. Safe startup pattern
```cpp
const int motorEnablePin = 5; // Example only
const int relayPin = 4;       // Example only

void setActuatorsOff() {
  analogWrite(motorEnablePin, 0);
  digitalWrite(relayPin, LOW); // Confirm actual relay OFF polarity
}

void setup() {
  pinMode(motorEnablePin, OUTPUT);
  pinMode(relayPin, OUTPUT);
  setActuatorsOff();
  Serial.begin(9600);
  Serial.println("Outputs initialized to configured safe state");
}

void loop() {
  // Start active behavior only after required checks pass.
}
```
This is a structure example, not a universal safe output configuration. Relay polarity and driver stop behavior must be verified on the actual hardware.

## 4. Fault and timeout logic
Possible fault conditions include invalid sensor values, controller disconnect, stale command, low supply detected by an appropriate sensor, or failed initialization. For each fault, document:
- How it is detected.
- Which state transition occurs.
- Which outputs are disabled or set to a known safe state.
- How the fault is reported.
- What human action is needed before restart.

A physical disconnect must remain available for motor and pump power.

## 5. Board and voltage compatibility
- Arduino Uno and ESP32 have different GPIO voltage limits; ESP32 GPIO is generally 3.3 V logic.
- A buck converter's output must be set and measured before connecting the load.
- Common ground is often required for signal reference between separately powered modules, but the exact grounding and isolation arrangement must follow the approved design.
- Do not connect 12 V directly to a 5 V or 3.3 V logic pin.
- Measure rails with a multimeter and disconnect power before changing wiring.
